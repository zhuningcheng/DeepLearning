```python
# 增强query将检索的文件和问题相关联，加强问题内容的相关性

def arguement(query,context=""):# context是给定的上下文
    # 提示词工程
    
    if not context:# 没有关联知识，模型任意生成回答
        return f"请回答下面的问题：{query}"
        
    else:# 有检索结果时，根据提示词来生成回答
        prompt = f"""请根据上下文来回答问题，如果上下文信息不在足以回答问题，请直接说“根据上下文信息，无法回答”
        上下文：{context}
        问题：{query}
"""
    return prompt
```

```python
# 生成回答
def generation(prompt):
    completion = client.chat.completions.create(
        model="deepseek-chat",
        messages=[
            {"role": "user", "content": prompt},
        ],
        stream=False
    )
    return completion.choices[0].message.content
```
